```python
input_data = df.drop(columns=['selling_price'])
output_data = df['selling_price']
```

```python
x_train, x_test, y_train, y_test = train_test_split(input_data, output_data, test_size=0.2, random_state=2)
```

```python
#model_creation
```

```python
model = LinearRegression()
# First, identify indices with NaN in x_train or y_train
nan_indices = x_train[x_train.isna().any(axis=1)].index.union(y_train[y_train.isna()].index)

# Drop those indices from both x_train and y_train
x_train = x_train.drop(nan_indices)
y_train = y_train.drop(nan_indices)
```

```python
print(len(x_train), len(y_train))
```

```python
# Drop rows with NaNs from both x_train and y_train
x_train = x_train.dropna().reset_index(drop=True)
y_train = y_train.dropna().reset_index(drop=True)

# Verify lengths
print("Length of x_train:", len(x_train))
print("Length of y_train:", len(y_train))

# Check if lengths now match
if len(x_train) == len(y_train):
    print("Datasets are aligned.")
else:
    # If they still don’t match, align explicitly by common indices
    common_indices = x_train.index.intersection(y_train.index)
    x_train = x_train.loc[common_indices].reset_index(drop=True)
    y_train = y_train.loc[common_indices].reset_index(drop=True)
```

```python
# Check for non-numeric values in x_train and y_train
print("Non-numeric values in x_train:")
print(x_train[~x_train.applymap(lambda x: isinstance(x, (int, float))).all(1)])

print("Non-numeric values in y_train:")
print(y_train[~y_train.apply(lambda x: isinstance(x, (int, float)))])
```

```python
# Replace empty strings or whitespace with NaN and then drop them
x_train = x_train.replace(r'^\s*$', float('nan'), regex=True).dropna().reset_index(drop=True)
y_train = y_train.replace(r'^\s*$', float('nan'), regex=True).dropna().reset_index(drop=True)
```

```python
# Verify lengths
print("Length of x_train:", len(x_train))
print("Length of y_train:", len(y_train))

# Fit the model if lengths match
if len(x_train) == len(y_train):
    model.fit(x_train, y_train)
else:
    print("Mismatch in lengths still exists.")
```

```python
# Align y_train with the indices of x_train
y_train = y_train.loc[x_train.index].reset_index(drop=True)
```

```python
# Check lengths again
print("Length of x_train:", len(x_train))
print("Length of y_train:", len(y_train))

# Fit the model if lengths match
if len(x_train) == len(y_train):
    model.fit(x_train, y_train)
else:
    print("There is still a length mismatch.")
```
